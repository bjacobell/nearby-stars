```python
full_target_list = df['Target'].values
full_cadence_id_list = df['Cadence ID'].values
full_h5_list = df['.h5 path'].values
full_dat_list = df['.dat path'].values
full_receiver_list = df['Receiver (X split)'].values
```

```python
cadences_info = []

for i in range(len(full_h5_list)//6):

    print(f'#{i+1} of {len(full_h5_list)//6} files ...')

    # for each set of 6 files, run FindEvent and PlotEvent

    h5_list = full_h5_list[6*i:6*i+6]
    dat_list = full_dat_list[6*i:6*i+6]

    # check that all h5 files in cadence exist and were searched
    # (not true for 3 files, 5 occurrences)
    #if 'None' in dat_list:
    #    continue

    target = full_target_list[6*i]
    cadence_id = full_cadence_id_list[6*i]
    receiver = full_receiver_list[6*i]
    s = dat_list[0].split('blc')
    if ('spliced' in dat_list[0]) & (s[1][2] == '_'):
        node = 'blc'+s[1][:2]
    else:
        node = os.path.basename(full_h5_list[6*i])[:5]

    # This block checks the cadence type:
    # Either ABACAD, BACADA, or None (if not a real cadence).
    # ABABAB is also checked, but does not occur in this dataset.
    #cadences = full_cadence_id_list[6*i:6*i+6]
    #receivers = np.unique(full_receiver_list[6*i:6*i+6])
    targets = []
    for j in range(6):
        t = df['.h5 path'].values[6*i+j].split('guppi_')[1][12:-21]
        if len(t) < 3:
            t = df['.h5 path'].values[6*i+j].split('guppi_')[1][6:-21] # account for GJ15B cadence listed with only integer MJD
        targets.append(t.upper()) # some stars not recorded in all uppercase
    odd_targets = np.unique(targets[::2])
    even_targets = np.unique(targets[1::2])
    cadence_type = 'None'
    off_check = 0
    for t in odd_targets:
        if 'OFF' in t:
            if len(even_targets) == 1:
                cadence_type = 'BACADA'
            off_check += 1
    for t in even_targets:
        if 'OFF' in t:
            if len(odd_targets) == 1:
                cadence_type = 'ABACAD'
            off_check += 1
    if (cadence_type == 'None') & (off_check == 0):
        n_odd_ons = len(np.unique(targets[::2]))
        n_even_ons = len(np.unique(targets[1::2]))
        if (n_odd_ons == 1) & (n_even_ons == 1):
            cadence_type = 'ABABAB'
        if (n_odd_ons == 1) & (n_even_ons != 1):
            cadence_type = 'ABACAD'
        if (n_odd_ons != 1) & (n_even_ons == 1):
            cadence_type = 'BACADA'

    #if not cadence_id == 4051:
    #    continue
    if cadence_type == 'ABACAD':
        target = targets[0]
    elif cadence_type == 'BACADA':
        target = targets[1]
    elif cadence_type == 'ABABAB':
        ### NEED TO SORT THIS OUT
        ### Add BOTH cadences and then figure out later which is real.
        ### This will mean rerunning the FindEvent code in the BABABA configuration for each of these cadences.
        ### Some will have both A and B in the BL star sample; some will have only one.

    #print(f'{cadence_id} {cadence_type} {target} {receiver} {targets}')

    #if cadence_type == 'None':
    #    continue

    #print(node)
    #print(dat_list[0])

    in_sample = 'Yes'

    additional_data = dfu.iloc[np.where(dfu['Name'].values == target)[0]]
    if len(additional_data) == 0:
        in_sample = 'No'
        print(f'{target} not found in dataframe!')
        cadences_info.append([
            target,
            np.nan,
            np.nan,
            np.nan,
            cadence_id,
            cadence_type,
            receiver,
            np.nan,
            np.nan,
            h5_list[0],
            h5_list[1],
            h5_list[2],
            h5_list[3],
            h5_list[4],
            h5_list[5],
            dat_list[0],
            dat_list[1],
            dat_list[2],
            dat_list[3],
            dat_list[4],
            dat_list[5]
        ])
        continue

    ra = additional_data['RA'].values[0]
    dec = additional_data['DEC'].values[0]
    vmag = additional_data['Vmag'].values[0]
    sptype = additional_data['Sp.Type'].values[0]
    dist = additional_data['dist (pc)'].values[0]

    cadences_info.append([
        target,
        ra,
        dec,
        dist,
        cadence_id,
        cadence_type,
        receiver,
        vmag,
        sptype,
        h5_list[0],
        h5_list[1],
        h5_list[2],
        h5_list[3],
        h5_list[4],
        h5_list[5],
        dat_list[0],
        dat_list[1],
        dat_list[2],
        dat_list[3],
        dat_list[4],
        dat_list[5]
    ])

cadence_keys = [
    'Target',
    'RA',
    'DEC',
    'Distance (pc)',
    'Cadence ID',
    'Cadence Type',
    'Receiver',
    'Vmag',
    'Spectral Type',
    '.h5 path 1',
    '.h5 path 2',
    '.h5 path 3',
    '.h5 path 4',
    '.h5 path 5',
    '.h5 path 6',
    '.dat path 1',
    '.dat path 2',
    '.dat path 3',
    '.dat path 4',
    '.dat path 5',
    '.dat path 6'
]

dfc = pd.DataFrame(data=cadences_info, columns=cadence_keys)
dfc
    # if (target == 'GJ412B') & (node == 'blc04') & (receiver == 'Rcvr1_2'):
    #     start_counter += 1
    
    # if start_counter == 0:
    #     continue
```
