```python
a = SkyCoord.from_name('HIP 42173')
print(a)
```

```python
gbt = EarthLocation.of_site('Green Bank Telescope')  # the easiest way... but requires internet
print(gbt)
#keck = EarthLocation.from_geodetic(lat=19.8283*u.deg, lon=-155.4783*u.deg, height=4160*u.m)
barycorr_start = a.radial_velocity_correction(obstime=Time('57750.20174', format='mjd'), location=gbt)
print(barycorr_start.to(u.km/u.s).value)
barycorr_end = a.radial_velocity_correction(obstime=Time(str(57750.20174+300/86400), format='mjd'), location=gbt)
print(barycorr_end.to(u.km/u.s).value)
#heliocorr = a.radial_velocity_correction('heliocentric', obstime=Time('57750.20174', format='mjd'), location=gbt)
#print(heliocorr.to(u.km/u.s))
```

```python
delv = barycorr_end.to(u.km/u.s).value - barycorr_start.to(u.km/u.s).value
print(delv/3e5 * 1538.712228e6 / 300) # should give barycenter-corrected drift rate in Hz/s
```
