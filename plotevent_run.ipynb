```python
import numpy as np 
import matplotlib.pyplot as plt 
import glob
import os
import pandas as pd
from turbo_seti.find_event.find_event_pipeline import find_event_pipeline
from turbo_seti.find_event.plot_event_pipeline import plot_event_pipeline
import multiprocessing
import logging
```

```python
event_csv_list = np.sort(glob.glob('/datax/scratch/benjb/bl_nearby_stars/seticore_output/findevent/*_trimmed.csv'))
print(len(event_csv_list))
for i in range(10):
    print(event_csv_list[i])
```

```python
data_loc = '/datax/scratch/benjb/bl_nearby_stars/seticore_output/plotevent/'

logger = logging.getLogger(__name__)

def process_files(xxx):

    n = int(xxx[0])
    batch_csv_list = xxx[1]

    for handler in logging.root.handlers[:]:
        logging.root.removeHandler(handler)
    logging.basicConfig(filename=f'{data_loc}/003_{n}_070725.log', filemode="a", level=logging.INFO)

    # with open(f'{data_loc}/000_{n}_061225.log') as f:
    #     lines = f.readlines()
    
    # skip_count = int((lines[-3].split('Searching #')[1]).split(' of ')[0])
    # logger.info('Skip count:', skip_count)

    for i, csv in enumerate(batch_csv_list):

        logger.info(f'Searching #{i+1} of {len(batch_csv_list)} files:')
        logger.info(f'{csv}')
        #if i+1 < skip_count:
        #    logger.info('Already plotted. Continuing ...')
        #    continue

        h5_loc = '/datax/scratch/benjb/bl_nearby_stars/seticore_output/findevent/'
        file_base = os.path.basename(csv).split('f3_trimmed')[0]

        #print(file_base)

        h5_list_path = h5_loc + file_base + 'h5_files.lst'

        df = pd.read_csv(csv)
        if len(df) == 0:
            logger.info('No events to plot in this file.')
            continue
        if len(df) > 100:
            logger.info(f'Too many events ({len(df)}) to plot in this file. Come back later!')
            continue

        logger.info(f'{len(df)} events in this file. Plotting ...')

        #if not os.path.exists(h5_list_path):
        #    print('DNE:', h5_list_path)

        try:
            plot_event_pipeline(csv, # full path of the CSV file built by find_event_pipeline()
                            h5_list_path, # full path of text file containing the list of .h5 files
                            filter_spec='f{}'.format(3), # filter threshold
                            user_validation=False, # Non-interactive
                            plot_dir=data_loc)
        except:
            logger.info(f'Error: HDF5 file(s) may be corrupted. Check: {h5_list_path}')
        
        logger.info(f'Plotting done.')
```

```python
data_loc = '/datax/scratch/benjb/bl_nearby_stars/seticore_output/plotevent/'

if __name__ == "__main__":
    # Define n+1 sets of files
    # The 1 accounts for the remainder after dividing N/n
    n = 31
    batch_size = len(event_csv_list)//n
    old_file_sets = [[str(i), event_csv_list[i*batch_size:(i+1)*batch_size]] for i in range(n)]
    old_file_sets.append([str(n), event_csv_list[n*batch_size:]])
    skip_files = []
    for i, batch in old_file_sets:
         with open(f'{data_loc}001_{i}_061725.log') as f:
            lines = f.readlines()
            for ii, l in enumerate(lines[:-4]):
                if 'Searching #' in l:
                    if not 'Error: ' in lines[ii+3]:
                        skip_files.append(lines[ii+1][14:-1]) # skip file if it was plotted successfully without throwing an error
    
    print(skip_files[0])
    old_file_csvs = np.concatenate([x[1] for x in old_file_sets])
    print(old_file_csvs[0])
    # subtract already-plotted files from the rest
    new_event_csv_list = list(set(old_file_csvs) - set(skip_files))
    new_batch_size = len(new_event_csv_list)//n

    file_sets = [[str(i), new_event_csv_list[i*new_batch_size:(i+1)*new_batch_size]] for i in range(n)]
    file_sets.append([str(n), new_event_csv_list[n*new_batch_size:]])

    # do it again for 001_ files
    skip_files = []
    for i, batch in file_sets:
         with open(f'{data_loc}002_{i}_062725.log') as f:
            lines = f.readlines()
            for ii, l in enumerate(lines[:-4]):
                if 'Searching #' in l:
                    if not 'Error: ' in lines[ii+3]:
                        skip_files.append(lines[ii+1][14:-1]) # skip file if it was plotted successfully without throwing an error
    
    print(skip_files[0])
    file_csvs_1 = np.concatenate([x[1] for x in file_sets])
    print(file_csvs_1[0])
    # subtract already-plotted files from the rest
    new_event_csv_list = list(set(file_csvs_1) - set(skip_files))
    new_batch_size = len(new_event_csv_list)//n

    file_sets = [[str(i), new_event_csv_list[i*new_batch_size:(i+1)*new_batch_size]] for i in range(n)]
    file_sets.append([str(n), new_event_csv_list[n*new_batch_size:]])

    # Create a pool of n+1 processes
    with multiprocessing.Pool(processes=n+1) as pool:
        # Map the process_files function to each file set
        pool.map(process_files, file_sets)

    print("All files processed.")
```
