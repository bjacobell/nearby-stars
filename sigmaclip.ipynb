```python
import numpy as np 
import matplotlib.pyplot as plt 
import blimpy as bl 
import glob
import os 
%matplotlib inline
```

```python
file = '/datax/scratch/benjb/single_coarse_guppi_59046_80036_DIAG_VOYAGER-1_0011.rawspec.0000.h5'
fb = bl.Waterfall(file)
fb.info()
```

```python
plt.plot(np.average(np.squeeze(fb.data), axis=0))
plt.yscale('log')
plt.show()
```

```python
powervals = np.average(np.squeeze(fb.data), axis=0)

plt.hist(np.log10(powervals), bins=100)
plt.yscale('log')
plt.show()
```

```python
powervals = fb.data.flatten()

plt.hist(np.log10(powervals))
plt.yscale('log')
plt.show()
```
